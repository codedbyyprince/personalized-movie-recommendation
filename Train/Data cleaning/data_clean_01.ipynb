```python
from collections import Counter

# Flatten all keyword names directly (cleaned to lowercase and stripped)
all_keywords = [d['name'].strip().lower() for sub in df['f_keywords'] for d in sub if 'name' in d]

# Count keyword frequencies
keywords_counter = Counter(all_keywords)

# Get top 150 most common keywords
common_keywords = [item[0] for item in keywords_counter]

# Replace rare keywords with 'other'
cleaned = []
for row in df['keywords_names']:
    # ensure row is a list and cleaned
    new_list = []
    for key in row:
        k = key.strip().lower()  # clean it before comparing
        if k in common_keywords:
            new_list.append(k)
        
    new_list = list(set(new_list))
    cleaned.append(new_list)

# Update DataFrame
df['keywords_names'] = cleaned
len(keywords_counter)
```

```python
from collections import Counter

# Flatten all keyword names directly (cleaned to lowercase and stripped)
all_keywords = [d['name'].strip().lower() for sub in df['f_keywords'] for d in sub if 'name' in d]

# Count keyword frequencies
keywords_counter = Counter(all_keywords)

# Get top 150 most common keywords
common_keywords = [item[0] for item in keywords_counter.most_common(200)]

# Replace rare keywords with 'other'
cleaned = []
for row in df['keywords_names']:
    # ensure row is a list and cleaned
    new_list = []
    for key in row:
        k = key.strip().lower()  # clean it before comparing
        if k in common_keywords:
            new_list.append(k)
        else:
            new_list.append('other')
    new_list = list(set(new_list))
    cleaned.append(new_list)

# Update DataFrame
df['keywords_names'] = cleaned
len(keywords_counter)
```
