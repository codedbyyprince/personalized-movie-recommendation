```python
import pandas as pd
import ast
import numpy as np
from sklearn.preprocessing import RobustScaler, MultiLabelBinarizer, OneHotEncoder
```

```python
df = pd.read_csv('/media/prince/5A4E832F4E83034D/Movie recomender/Train_v2/v3_cleaneed.csv')
```

```python
num_cols = ['popularity','runtime','vote_average','vote_count']

scaler = RobustScaler()
df[num_cols] = scaler.fit_transform(df[num_cols])
```

```python
multi_cols = ['genre_names','keywords_names', 'comapanies_categories']
one_hot = ['original_language']

mlb = MultiLabelBinarizer()
ohe = OneHotEncoder()

```

```python
for col in ['genre_names', 'keywords_names', 'companies_categories']:
    df[col] = df[col].apply(ast.literal_eval)

for col in ['genre_names', 'keywords_names', 'companies_categories']:
    mlb = MultiLabelBinarizer()
    df = df.join(pd.DataFrame(mlb.fit_transform(df[col]), columns=mlb.classes_))

df = df.drop(['genre_names', 'keywords_names', 'companies_categories'], axis=1)

# --- One Hot Encoding ---
df = pd.get_dummies(df, columns=['original_language'])
```
